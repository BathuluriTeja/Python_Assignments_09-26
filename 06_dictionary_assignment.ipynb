```python
print("\nQuestion 43: Create a dictionary with nested lists: {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}")
nested_lists = {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}
print(nested_lists)
```

```python
for key, val in nested_lists.items():
    if isinstance(val[0], list):  # matrix case
        print(key, ":", [sum(sublist) for sublist in val])
    else:  # vector case
        print(key, ":", sum(val))
```
