```python
# Question 1: Product of first 10 natural numbers
print("Question 1: Product of first 10 natural numbers")
product = 1
for i in range(1, 11):
    product = product * i
print("Product:", product)
```

```python
# Question 2: Remainder when 156 is divided by 7
print("\nQuestion Remainder when 156 is divided by 7")
remainder = 156 % 7
print("Remainder:", remainder)
```

```python
# Question 3: Square of 25
print("\nQuestion 3: Square of 25")
square = 25 ** 2
print("Square:", square)
```

```python
# Question 4: Cube root of 125
print("\nQuestion 4: Cube root of 125")
cube_root = round(125 ** (1/3))
print("Cube root:", cube_root)
```

```python
# Question 5: Sum of digits in 12345
print("\nQuestion 5: Sum of digits in number 12345")
number = 12345
digit_sum = 0
while number > 0:
    digit_sum += number % 10
    number = number // 10
print("Sum of digits:", digit_sum)
```

```python
# Question 6: Check if 97 is a prime number
print("\nQuestion 6: Check if 97 is a prime number")
number = 97
is_prime = True
for i in range(2, int(number ** 0.5) + 1):
    if number % i == 0:
        is_prime = False
        break

if is_prime:
    print("97 is a prime number")
else:
    print("97 is not a prime number")
```

```python
# Question 7: Factorial of 8
print("\nQuestion 7: Find the factorial of 8")
factorial = 1
for i in range(1, 9):
    factorial *= i
print("Factorial:", factorial)
```

```python
# Question 8: Average of given numbers
print("\nQuestion 8: Average of numbers: 15, 23, 31, 42, 56")
numbers = [15, 23, 31, 42, 56]
average = sum(numbers) / len(numbers)
print("Average:", average)
```

```python
# Question 9: GCD of 48 and 36
print("\nQuestion 9: GCD of 48 and 36")
a = 48
b = 36
while b != 0:
    a, b = b, a % b
print("GCD:", a)
```

```python
# Question 10: Sum of first 20 odd numbers
print("\nQuestion 10: Sum of first 20 odd numbers")
total = 0
for i in range(1, 40, 2):
    total += i
print("Sum:", total)
```
